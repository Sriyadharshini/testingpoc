```python
df = spark.read.format("delta").load(
    "abfss://bronze@stcpipeinedev.dfs.core.windows.net/customer"
)
```

```python
df = df.dropDuplicates(["customer_id"])
df = df.filter("customer_id IS NOT NULL")
```

```python
df = df.toDF(*[c.lower().strip() for c in df.columns])
```

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .save("abfss://silver@stcpipeinedev.dfs.core.windows.net/customer")
```
