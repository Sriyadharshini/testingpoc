```python
from pyspark.sql.functions import current_timestamp, col

df = spark.read.parquet(
    "abfss://staging@stcpipeinedev.dfs.core.windows.net/customer.parquet"
)

df = (df
      .withColumn("ingestion_time", current_timestamp())
      .withColumn("source_file", col("_metadata.file_path"))
)

df.write \
    .format("delta") \
    .mode("overwrite") \
    .save("abfss://bronze@stcpipeinedev.dfs.core.windows.net/customer")
```
